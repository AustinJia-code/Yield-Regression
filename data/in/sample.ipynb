```python
import pandas as pd
from pathlib import Path

def create_mini_csv(input_path, n_rows=100, random_state=42):
    """
    Takes n random rows from a CSV and saves to {filename}_mini.csv
    
    Args:
        input_path: Path to input CSV
        n_rows: Number of rows to sample (default 100)
        random_state: Random seed for reproducibility
    """
    # Read CSV
    df = pd.read_csv(input_path)
    print(f"Original CSV: {len(df)} rows")
    
    # Sample random rows
    n_sample = min(n_rows, len(df))  # Don't sample more than available
    df_mini = df.sample(n=n_sample, random_state=random_state)
    print(f"Sampled: {len(df_mini)} rows")
    
    # Create output path
    input_path = Path(input_path)
    output_path = input_path.parent / f"{input_path.stem}_mini{input_path.suffix}"
    
    # Save
    df_mini.to_csv(output_path, index=False)
    print(f"Saved to: {output_path}")
    
    return output_path

# Example usage
if __name__ == "__main__":
    # Create mini versions of your data files
    TRAIN_PATH = "./cattle_data_train.csv"
    create_mini_csv(TRAIN_PATH, n_rows=100)
    
    print("\nDone! Use these paths:")
    print(f"TRAIN_PATH_MINI = '{Path(TRAIN_PATH).parent / Path(TRAIN_PATH).stem}_mini.csv'")
```
